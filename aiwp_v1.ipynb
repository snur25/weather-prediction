```python
import cdsapi
import xarray as xr
import os

URL = "https://cds.climate.copernicus.eu/api"
from google.colab import userdata
KEY = userdata.get("CDS_API_KEY")

client = cdsapi.Client(url=URL, key=KEY)

# 1. Loop through each year one by one to bypass the server limits
years = range(2010, 2025)

for year in years:
    filename = f'konya_peak_daily_{year}.nc'

    # 2. Check if file already exists so we don't redownload it if Colab restarts!
    if not os.path.exists(filename):
        print(f"Downloading data for {year}...")
        client.retrieve(
            'reanalysis-era5-single-levels',
            {
                'product_type': 'reanalysis',
                'format': 'netcdf',
                'variable': [
                    '10m_u_component_of_wind',
                    '10m_v_component_of_wind',
                    '2m_temperature',
                    'mean_sea_level_pressure'
                ],
                'year': str(year), # Requesting just this specific year
                'month': [str(m).zfill(2) for m in range(1, 13)],
                'day': [str(d).zfill(2) for d in range(1, 32)],
                'time': ['12:00'],
                'area': [38.5, 31.5, 37.0, 33.5],
            },
            filename
        )
    else:
        print(f"File {filename} already exists, skipping download.")

print("All years successfully downloaded!")

# 3. Seamlessly stitch all 16 separate files into one continuous timeline
ds = xr.open_mfdataset('konya_peak_daily_*.nc', combine='by_coords')

print(ds)
```
