```python
import cdsapi
import xarray as xr
import os

URL = "https://cds.climate.copernicus.eu/api"
from google.colab import userdata
KEY = userdata.get("CDS_API_KEY")

client = cdsapi.Client(url=URL, key=KEY)

# 1. Loop through each year one by one to bypass the server limits
years = range(2010, 2025)

for year in years:
    filename = f'konya_peak_daily_{year}.nc'

    # 2. Check if file already exists so we don't redownload it if Colab restarts!
    if not os.path.exists(filename):
        print(f"Downloading data for {year}...")
        client.retrieve(
            'reanalysis-era5-single-levels',
            {
                'product_type': 'reanalysis',
                'format': 'netcdf',
                'variable': [
                    '10m_u_component_of_wind',
                    '10m_v_component_of_wind',
                    '2m_temperature',
                    'mean_sea_level_pressure'
                ],
                'year': str(year), # Requesting just this specific year
                'month': [str(m).zfill(2) for m in range(1, 13)],
                'day': [str(d).zfill(2) for d in range(1, 32)],
                'time': ['12:00'],
                'area': [38.5, 31.5, 37.0, 33.5],
            },
            filename
        )
    else:
        print(f"File {filename} already exists, skipping download.")

print("All years successfully downloaded!")

# 3. Seamlessly stitch all 16 separate files into one continuous timeline
ds = xr.open_mfdataset('konya_peak_daily_*.nc', combine='by_coords')

print(ds)
```

```python
# Cell 1 - Imports, configuration and hardware information
import glob
import os
import platform
import threading
import time

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import psutil
import xarray as xr
from sklearn.linear_model import Ridge
from sklearn.model_selection import TimeSeriesSplit
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler

SEED = 42
np.random.seed(SEED)

# Folders on Google Drive
DATA_DIR = "/content/drive/MyDrive/era5_data"
PROJECT_DIR = "/content/drive/MyDrive/aiwp_project"
RESULTS_DIR = os.path.join(PROJECT_DIR, "results")
FIG_DIR = os.path.join(PROJECT_DIR, "figures")

# Experiment settings
LOCATIONS = {"Konya": (37.87, 32.48), "Karapinar": (37.72, 33.55)}
HORIZON_DAYS = 1       # forecast the 12:00 UTC temperature this many days ahead
N_SPLITS = 5           # number of time series cross-validation folds
TEST_SIZE_DAYS = 730   # each test fold covers about two years
INFER_REPEATS = 20     # repeat predictions to time them reliably


def hardware_info():
    """Describe the machine, so every timing in the thesis can be reproduced."""
    info = {
        "python": platform.python_version(),
        "cpu_cores": os.cpu_count(),
        "ram_gb": round(psutil.virtual_memory().total / 1e9, 1),
        "gpu": "none",
    }
    try:
        import torch
        if torch.cuda.is_available():
            info["gpu"] = torch.cuda.get_device_name(0)
    except ImportError:
        pass
    return info


HARDWARE = hardware_info()
print("Hardware:", HARDWARE)
```

```python
# Cell 2 - Load the ERA5 files from Google Drive
from google.colab import drive

drive.mount("/content/drive")

for folder in (PROJECT_DIR, RESULTS_DIR, FIG_DIR):
    os.makedirs(folder, exist_ok=True)

files = sorted(glob.glob(os.path.join(DATA_DIR, "konya_peak_daily_*.nc")))
if not files:
    raise FileNotFoundError(f"No konya_peak_daily_*.nc files found in {DATA_DIR}")
print(f"Found {len(files)} files: {os.path.basename(files[0])} ... {os.path.basename(files[-1])}")

ds = xr.open_mfdataset(files, combine="by_coords")
ds = ds.drop_vars(["number", "expver"], errors="ignore")
if "time" in ds.dims and "valid_time" not in ds.dims:
    ds = ds.rename({"time": "valid_time"})
ds = ds.sortby("valid_time").load()  # the data is tiny (about 6 MB), so keep it in memory

missing_vars = {"t2m", "u10", "v10", "msl"} - set(ds.data_vars)
if missing_vars:
    raise KeyError(f"Missing variables in the dataset: {missing_vars}")

print(dict(ds.sizes))
print("Period:", str(ds["valid_time"].values[0])[:10], "to", str(ds["valid_time"].values[-1])[:10])
```

```python
# Cell 3 - Extract one daily table per location (nearest ERA5 grid cell)
def extract_point_frame(dataset, lat, lon):
    """Return a daily DataFrame for the grid cell nearest to (lat, lon)."""
    point = dataset.sel(latitude=lat, longitude=lon, method="nearest")
    used_lat = float(point["latitude"])
    used_lon = float(point["longitude"])

    dates = pd.DatetimeIndex(dataset["valid_time"].values).normalize()
    domain = dataset[["u10", "v10", "msl"]].mean(dim=["latitude", "longitude"])

    frame = pd.DataFrame(
        {
            "t2m_c": point["t2m"].values - 273.15,       # Kelvin -> Celsius
            "u10": point["u10"].values,
            "v10": point["v10"].values,
            "msl_hpa": point["msl"].values / 100.0,       # Pa -> hPa
            "dom_u10": domain["u10"].values,              # mean over the whole downloaded box
            "dom_v10": domain["v10"].values,
            "dom_msl_hpa": domain["msl"].values / 100.0,
        },
        index=dates,
    )
    frame.index.name = "date"
    frame["wind_speed"] = np.hypot(frame["u10"], frame["v10"])
    return frame, used_lat, used_lon


frames = {}
for name, (lat, lon) in LOCATIONS.items():
    frame, used_lat, used_lon = extract_point_frame(ds, lat, lon)

    if not frame.index.is_unique or not frame.index.is_monotonic_increasing:
        raise ValueError(f"{name}: dates are duplicated or unsorted")
    full_range = pd.date_range(frame.index[0], frame.index[-1], freq="D")
    if len(full_range) != len(frame):
        raise ValueError(f"{name}: {len(full_range) - len(frame)} daily values are missing")

    frames[name] = frame
    print(f"{name}: requested ({lat}, {lon}) -> grid cell ({used_lat}, {used_lon}), {len(frame)} days")

frames["Konya"].head()
```

```python
# Cell 4 - Data quality checks and first plots
for name, frame in frames.items():
    print(f"\n=== {name} ===")
    print("Missing values per column:", int(frame.isna().sum().sum()))
    print(frame[["t2m_c", "wind_speed", "msl_hpa"]].describe().round(2))
    if not frame["t2m_c"].between(-40, 55).all():
        raise ValueError(f"{name}: temperatures outside the plausible range")
    if not frame["msl_hpa"].between(950, 1060).all():
        raise ValueError(f"{name}: pressures outside the plausible range")

fig, axes = plt.subplots(2, 1, figsize=(11, 7))

axes[0].plot(frames["Konya"].index, frames["Konya"]["t2m_c"], linewidth=0.6)
axes[0].set_title("Konya: 12:00 UTC 2 m temperature, 2010-2024")
axes[0].set_ylabel("Temperature (°C)")

for name, frame in frames.items():
    seasonal = frame["t2m_c"].groupby(frame.index.dayofyear).mean()
    axes[1].plot(seasonal.index, seasonal.values, label=name)
axes[1].set_title("Average seasonal cycle (mean by day of year)")
axes[1].set_xlabel("Day of year")
axes[1].set_ylabel("Temperature (°C)")
axes[1].legend()

plt.tight_layout()
fig.savefig(os.path.join(FIG_DIR, "01_data_overview.png"), dpi=150)
plt.show()
```
